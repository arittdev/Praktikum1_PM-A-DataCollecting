```python
#202531034_HaritsNakhlahPutra
import pandas as pd # library untuk pengolahan data tabel
import numpy as np # library untuk matematikan dan matriks
import matplotlib.pyplot as plt # library untuk grafik dan visual
import seaborn as sns # library untuk desain dan grafis
```

```python
#202531034_HaritsNakhlahPutra
from sklearn import datasets
from skimage import data
from skimage.color import rgb2gray
from skimage.transform import resize

#1. Memanggil dataset
data_titanic = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df_csv = pd.read_csv(data_titanic)
df_csv.head()
```

```python
#202531034_HaritsNakhlahPutra
image = data.chelsea()
gray_image = rgb2gray(image)
pixelated = resize(gray_image[50:250, 150:350], (15,15), anti_aliasing = True)
pixelated_255 = (pixelated * 255).astype(int)

fig, ax = plt.subplots(1, 2, figsize = (10, 4))
ax[0].imshow(image)
ax[0].set_title(f"Gambar Asli Yang Dilihat Manusia: {image.shape}")
ax[1].imshow(pixelated_255, cmap='gray')
ax[1].set_title("Gambar Yang Dilihat Oleh Komputer")

for i in range (15):
  for j in range (15):
    val = pixelated_255[i,j]
    color = "white" if val < 128 else "black"
    ax[1].text(j, i, val, ha="center", color = color, fontsize = 7)

plt.tight_layout()
plt.show()
```
